```sql
-- GoodCabs Gold layer validation
-- Run after gold_deploy.sql

SELECT *
FROM transportation.gold.fact_trips
LIMIT 10;

SELECT COUNT(*) AS total_trips
FROM transportation.gold.fact_trips;

USE CATALOG transportation;
USE SCHEMA gold;
SHOW VIEWS IN gold;

SELECT
    COUNT(*) AS total_trips,
    COUNT(DISTINCT city_id) AS total_cities,
    ROUND(SUM(sales_amt), 2) AS total_revenue
FROM transportation.gold.fact_trips;

SELECT table_name
FROM transportation.information_schema.views
WHERE table_schema = 'gold'
ORDER BY table_name;

SELECT
    city_name,
    COUNT(*) AS total_trips,
    ROUND(SUM(sales_amt), 2) AS total_revenue,
    ROUND(AVG(passenger_rating), 2) AS avg_rating
FROM transportation.gold.fact_trips
GROUP BY city_name
ORDER BY total_revenue DESC;
```
